```python
# ============================
# PHASE 1: VIDEO FEATURE EXTRACTION
# ============================

# Install once if needed
# !pip install -q transformers timm accelerate einops opencv-python decord

import torch
import cv2
import numpy as np
from pathlib import Path
from transformers import VideoMAEImageProcessor, VideoMAEModel

# ----------------------------
# DEVICE CHECK
# ----------------------------

print("=" * 50)
print("DEVICE INFO")
print("=" * 50)

print("Torch:", torch.__version__)
print("CUDA:", torch.cuda.is_available())

device = "cuda" if torch.cuda.is_available() else "cpu"

if device == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))
    print(
        "VRAM:",
        round(
            torch.cuda.get_device_properties(0).total_memory / 1024**3,
            2
        ),
        "GB"
    )

# ----------------------------
# LOAD MODEL
# ----------------------------

print("\nLoading VideoMAE...")

processor = VideoMAEImageProcessor.from_pretrained(
    "MCG-NJU/videomae-base"
)

model = VideoMAEModel.from_pretrained(
    "MCG-NJU/videomae-base"
).to(device)

model.eval()

print("VideoMAE Loaded")

# ----------------------------
# VIDEO PATH
# ----------------------------

video_path = r"C:/Users/ravur/Downloads/1b3df5a029af.mp4"

assert Path(video_path).exists(), f"Video not found: {video_path}"

# ----------------------------
# READ VIDEO
# ----------------------------

cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
duration = total_frames / fps

print("\nVIDEO INFO")
print("FPS:", fps)
print("Frames:", total_frames)
print("Duration:", round(duration, 2), "seconds")

# ----------------------------
# SAMPLE 16 FRAMES
# ----------------------------

indices = np.linspace(
    0,
    total_frames - 1,
    16,
    dtype=int
)

frames = []

for idx in indices:

    cap.set(cv2.CAP_PROP_POS_FRAMES, idx)

    success, frame = cap.read()

    if success:
        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        frames.append(frame)

cap.release()

print("\nExtracted Frames:", len(frames))
print("Frame Shape:", frames[0].shape)

# ----------------------------
# FEATURE EXTRACTION
# ----------------------------

inputs = processor(
    frames,
    return_tensors="pt"
)

pixel_values = inputs["pixel_values"].to(device)

print("\nPixel Values Shape:", pixel_values.shape)

with torch.no_grad():

    outputs = model(
        pixel_values=pixel_values
    )

features = outputs.last_hidden_state

print("\nSUCCESS")
print("Feature Shape:", features.shape)

print("\nPhase 1 Completed")
```

```python
import cv2
import torch
import numpy as np

from transformers import VideoMAEImageProcessor
from transformers import VideoMAEModel
from sklearn.metrics.pairwise import cosine_similarity

video_path = r"C:\Users\ravur\PERSONAL\Desktop\BVRIT-COLLEGE\MiniProject\POV_Video_Generation.mp4"

device = "cuda"

processor = VideoMAEImageProcessor.from_pretrained(
    "MCG-NJU/videomae-base"
)

model = VideoMAEModel.from_pretrained(
    "MCG-NJU/videomae-base"
).to(device)

model.eval()

# --------------------------------
# LOAD VIDEO
# --------------------------------

cap = cv2.VideoCapture(video_path)

all_frames = []

while True:

    ok, frame = cap.read()

    if not ok:
        break

    frame = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2RGB
    )

    all_frames.append(frame)

cap.release()

print("Total Frames:", len(all_frames))

# --------------------------------
# CREATE WINDOWS OF 16 FRAMES
# --------------------------------

window_size = 16

windows = []

for i in range(
    0,
    len(all_frames) - window_size + 1,
    window_size
):
    windows.append(
        all_frames[i:i+window_size]
    )

print("Windows:", len(windows))

# --------------------------------
# EXTRACT EMBEDDINGS
# --------------------------------

embeddings = []

for idx, window in enumerate(windows):

    inputs = processor(
        window,
        return_tensors="pt"
    )

    pixel_values = (
        inputs["pixel_values"]
        .to(device)
    )

    with torch.no_grad():

        outputs = model(
            pixel_values=pixel_values
        )

    emb = (
        outputs.last_hidden_state
        .mean(dim=1)
        .squeeze()
        .cpu()
        .numpy()
    )

    embeddings.append(emb)

print(
    "Embeddings Extracted:",
    len(embeddings)
)

# --------------------------------
# TEMPORAL SIMILARITIES
# --------------------------------

if len(embeddings) >= 3:

    early = embeddings[0]
    mid = embeddings[len(embeddings)//2]
    late = embeddings[-1]

    print("\nTemporal Similarities")

    print(
        "Early-Mid:",
        cosine_similarity(
            [early],
            [mid]
        )[0][0]
    )

    print(
        "Mid-Late:",
        cosine_similarity(
            [mid],
            [late]
        )[0][0]
    )

    print(
        "Early-Late:",
        cosine_similarity(
            [early],
            [late]
        )[0][0]
    )

else:

    print(
        "\nVideo too short."
    )
```
